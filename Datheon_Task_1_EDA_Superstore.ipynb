```python
Q1 = sales_df["Sales"].quantile(0.25)
Q3 = sales_df["Sales"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

```python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
```

```python
sales_outliers = sales_df[
    (sales_df["Sales"] < lower_limit) |
    (sales_df["Sales"] > upper_limit)
]

print("Total Outliers:", len(sales_outliers))
print("Outlier Percentage:", round(
    len(sales_outliers) / len(sales_df) * 100, 2
), "%")
```

```python
plt.figure(figsize=(9, 5))

plt.boxplot(sales_df["Sales"])

plt.title("Box Plot - Sales Outlier Detection")
plt.ylabel("Sales")

plt.show()
```

```python
Q1_profit = sales_df["Profit"].quantile(0.25)
Q3_profit = sales_df["Profit"].quantile(0.75)

IQR_profit = Q3_profit - Q1_profit

print("Q1:", Q1_profit)
print("Q3:", Q3_profit)
print("IQR:", IQR_profit)
```

```python
lower_limit_profit = Q1_profit - 1.5 * IQR_profit
upper_limit_profit = Q3_profit + 1.5 * IQR_profit

print("Lower Limit:", lower_limit_profit)
print("Upper Limit:", upper_limit_profit)
```

```python
profit_outliers = sales_df[
    (sales_df["Profit"] < lower_limit_profit) |
    (sales_df["Profit"] > upper_limit_profit)
]

print("Total Profit Outliers:", len(profit_outliers))

print("Outlier Percentage:", round(
    len(profit_outliers) / len(sales_df) * 100, 2
), "%")
```

```python
plt.figure(figsize=(9, 5))

plt.boxplot(sales_df["Profit"])

plt.title("Box Plot - Profit Outlier Detection")
plt.ylabel("Profit")

plt.show()
```
